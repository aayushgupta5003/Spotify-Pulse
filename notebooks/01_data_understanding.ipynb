```python
streaming_history = pd.read_csv("../data/raw/spotify_streaming_history.csv")

user_behavior = pd.read_csv("../data/raw/spotify_user_behavior.csv")

user_preferences = pd.read_excel("../data/raw/spotify_user_preferences.xlsx")
```

```python
# Checking dimensions

print("Streaming History:", streaming_history.shape)
print("User Behavior:", user_behavior.shape)
print("User Preferences:", user_preferences.shape)
```

```python
# =====================================================
# Dataset Overview
# =====================================================

datasets = {
    "Streaming History": streaming_history,
    "User Behavior": user_behavior,
    "User Preferences": user_preferences
}

for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("=" * 60)
    print(f"Shape: {df.shape}")
    print("\nColumns:")
    print(list(df.columns))
    print("\n")
```
