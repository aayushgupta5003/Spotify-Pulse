```python
def data_quality_report(df, name):
    print("=" * 70)
    print(f"DATA QUALITY REPORT : {name}")
    print("=" * 70)

    print(f"\nShape : {df.shape}")

    print("\nColumn Names:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\nMemory Usage:")
    print(df.memory_usage(deep=True).sum() / (1024**2), "MB")
```

```python
data_quality_report(preferences_df, "User Preferences")
```

```python
def missing_value_report(df):
    report = pd.DataFrame({
        "Column": df.columns,
        "Missing Values": df.isnull().sum().values,
        "% Missing": (df.isnull().sum()/len(df)*100).round(2).values
    })

    return report.sort_values("% Missing", ascending=False)
```

```python
missing_value_report(streaming_df)
```

```python
missing_value_report(behavior_df)
```

```python
missing_value_report(preferences_df)
```

```python
def duplicate_report(df):
    duplicates = df.duplicated().sum()

    print(f"Duplicate Rows : {duplicates}")

    if duplicates > 0:
        display(df[df.duplicated()].head())
```

# Data Cleaning

Cleaning decisions:
- Remove duplicate rows from Streaming History.
- Remove duplicate rows from User Preferences.
- Keep User Behavior unchanged.
- Fill missing values in `reason_start` and `reason_end` with `"unknown"`.
- Preserve structural missing values in the User Preferences dataset.

```python
# -----------------------------
# Remove duplicate rows
# -----------------------------

streaming_df = streaming_df.drop_duplicates()

preferences_df = preferences_df.drop_duplicates()
```

```python
# -----------------------------
# Fill missing values
# -----------------------------

streaming_df["reason_start"] = streaming_df["reason_start"].fillna("unknown")

streaming_df["reason_end"] = streaming_df["reason_end"].fillna("unknown")
```

```python
print("Streaming Shape :", streaming_df.shape)
print("Behavior Shape  :", behavior_df.shape)
print("Preferences Shape :", preferences_df.shape)

print("\nMissing Values in Streaming")
print(streaming_df[["reason_start","reason_end"]].isnull().sum())

print("\nDuplicate Rows")
print(streaming_df.duplicated().sum())
print(behavior_df.duplicated().sum())
print(preferences_df.duplicated().sum())
```

```python
# ---------------------------------------
# Create Tracks Dimension Table
# ---------------------------------------

tracks_df = streaming_df[
    ["spotify_track_uri", "track_name", "artist_name", "album_name"]
].drop_duplicates().reset_index(drop=True)
```

```python
tracks_df.insert(
    0,
    "track_id",
    range(1, len(tracks_df) + 1)
)
```
